```python
import numpy as np
```

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
# Create a synthetic dataset simulating monthly sales and profit
```

```python
np.random.seed(42)
```

```python
months= ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
```

```python
sales=np.random.randint(15000,35000, size=12)
```

```python
profit= sales*np.random.uniform(0.1, 0.3, size=12)
```

```python
df= pd.DataFrame({'Month':months, 'Sales':sales, 'Profit':np.round(profit, 2)})
```

```python
df
```

```python
plt.figure(figsize=(10, 5))
```

```python
plt.plot(df['Month'], df['Sales'], marker='o', color='b', label='Sales ($)', linewidth=2)
```

```python
plt.plot(df['Month'], df['Profit'], marker='s', color='g', label='Profit ($)', linewidth=2)
```

```python
plt.title('2026 Monthly Sales vs Profit Performance', fontsize=14)
```

```python
plt.xlabel('Month',fontsize=12)
```

```python
plt.ylabel('Amount ($)', fontsize=12)
```

```python
plt.grid(True, linestyle='--', alpha=0.6)
```

```python
plt.legend()
```

```python
plt.tight_layout()
```

```python
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df['Month'], df['Sales'], marker='o', label='Sales' ,linewidth=2)
plt.plot(df['Month'], df['Profit'], marker='s', label='Profit', linewidth=2)
plt.title('Monthly Sales and Profit', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Amount ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6) 
plt.legend()
plt.show()
```
